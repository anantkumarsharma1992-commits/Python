```python
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [120, 150, 170, 140, 200, 230]
```

```python
fig= px.bar(x= months,y= sales, title="Monthly Sales", labels={"x": "Months","y": "Sales"})
fig.show()
```

```python
# # 2Line Chart

# Represent the same monthly sales as a line chart.

fig= px.line(x=months, y=sales, title="Monthly Sales",labels={"x":"Months","y": "Sales in $k"})
fig.show()
```

```python
# Multiple Categories

# Create a chart comparing sales of

# Laptop

# Mobile

# Tablet

# # across six months

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

laptop_sales = [150, 170, 160, 180, 200, 220]
mobile_sales = [300, 280, 320, 350, 340, 400]
tablet_sales = [90, 100, 95, 110, 130, 125]
```

```python
fig= go.Figure(data=[
    go.Bar(x=months, y= laptop_sales, name="Laptop"),
    go.Bar(x=months, y= mobile_sales, name="Mobile"),
    go.Bar(x=months, y= tablet_sales, name="Tablet")

])

fig.update_layout(title= "Category wise sales data", xaxis_title= "Months",yaxis_title= "Sales")

fig.show()
```

```python
# Dashboard Thinking Challenge

# Create at least four separate charts representing a fictional company's:

# Revenue

# Expenses

# Employees

# Customers

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue = [199700, 212100, 236700, 238600, 265400, 245400, 260100, 277700, 313600, 326500, 341600, 373200]

expenses = [117800, 143300, 143700, 169300, 185300, 169500, 172200, 165400, 182900, 188300, 208500, 239400]

employees = [42, 42, 45, 45, 48, 50, 50, 53, 55, 55, 58, 60]

customers = [1199, 1411, 1555, 1877, 1903, 2072, 2519, 2519, 2619, 2732, 3316, 3316]
```

```python
#Line chart depicting no of cusotmers every month

fig= px.line(x= months, y= customers, title= "MOnthly Customers", labels={"x": "Months", "y": "Customers"})
fig.show()

# The no of customers increases every month, and as expected they are highest in Year end to holidays
```

```python
#Line chart depicting no of cusotmers every month

fig= px.bar(x= months, y= employees, title= "Employee count per month", labels={"x": "Months", "y": "Employee"})
fig.show()
# We have more number of employees in Nov and Dec to handle year end pressure 
```

```python
fig= go.Figure(data=[
    go.Bar(x=months, y= revenue, name="Revenue"),
    go.Bar(x=months, y= expenses, name="Expenses"),
    

])

fig.update_layout(title= "Monthly Revenue and Expense data", xaxis_title= "Months",yaxis_title= "Sales in $k")

fig.show()

#We were in profit every month as montly revenue is more than expenses 
```

```python
profit= np.array(revenue)- np.array(expenses)

fig= px.line(x=months, y= profit, title= "Monthly Profit", labels={"x": "Month", "y": "Profit"})
fig.show()
```
