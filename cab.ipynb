```python
class vehicle:
    def __init__(self, veh_no, brand, driver_name,priceperkm):
        self.vehical_no= veh_no
        self.brand= brand
        self.driver_name= driver_name
        self.priceperkm= priceperkm

   
```

```python
class bike(vehicle):
    
    def __init__(self, veh_no, brand, driver_name,priceperkm):
        super(). __init__(veh_no,brand,driver_name,priceperkm)

    def display_info(self,distance):
        fare= distance* self.priceperkm
        print(f"the vehicle has been book with {self.vehical_no} from brand {self.brand}. The driver is {self.driver_name} and the price per km will be {self.priceperkm}. Your total fare is {fare}")

    
```

```python
class car(vehicle):
    
    def __init__(self, veh_no, brand, driver_name,priceperkm):
        super(). __init__(veh_no,brand,driver_name,priceperkm)

    def display_info(self,distance):
        fare= distance* self.priceperkm
        print(f"the vehicle has been book with {self.vehical_no} from brand {self.brand}. The driver is {self.driver_name} and the price per km will be {self.priceperkm}. Your total fare is {fare}")

    
```

```python
car1 = car("MH12AB1234", "Toyota", "Rahul", 20)
car2 = car("MH14XY5678", "Honda", "Amit", 25)

bike1 = bike("MH13CD4321", "Yamaha", "Suresh", 10)
bike2 = bike("MH15EF8765", "Bajaj", "Ramesh", 12)
```

```python
vehicle={1:car1,2:car2,3:bike1,4:bike2}
```

```python
##interaction with user

print("welcome to fare calculator")
print("select a vehicle:")
print("1. Car - Toyota (Rahul)")
print("2. Car - Honda (Amit)")
print("3. Bike - Yamaha (Suresh)")
print("4. Bike - Bajaj (Ramesh)")

n= int(input("Enter your choice:"))
dist= int(input("Enter travel distance:"))

try:
    if n in vehicle:
        selected_vehicle= vehicle[n]
        selected_vehicle.display_info(dist)
    else:
        print("Invalid choice, please select again")
except Exception as e:
    print("error occure:",e)
    
```
