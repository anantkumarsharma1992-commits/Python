```python
# Using Python file handling
# Read the entire log.
# Count total log entries.
# Count ERROR occurrences.
# Count WARNING occurrences.
# # Count INFO occurrences.
# Create
# errors.txt
# warnings.txt
# Store appropriate log entries inside those files.
# Bonus: Load the log information into Pandas and perform frequency analysis.
```

```python
with open("application.log",'r') as file:
    lines= file.readlines()
```

```python
lines
```

```python
# Count total log entries.
len(lines)
```

```python
error_data=[]
info_data=[]
warning_data=[]

for i in lines:
    if i.startswith("INFO"):
        info_data.append(i)
    if i.startswith("ERROR"):
        error_data.append(i)
    if i.startswith("WARNING"):
        warning_data.append(i)
```

```python
error_data
```

```python
info_data
```

```python
#  Count ERROR occurrences.
print(f"total number of error occurences is: {len(error_data)}")
```

```python
# Count WARNING occurrences.
print(f"total number of warning occurences is: {len(warning_data)}")
```

```python
# # Count INFO occurrences.
print(f"total number of info occurences is: {len(info_data)}")
```

```python
# errors.txt
error_file= open("error.txt", 'w')
for i in error_data:
    error_file.write(i + "\n")
error_file.close()
```

```python
# warning.txt

warning_file= open("warning.txt",'w')
for i in warning_data:
    warning_file.write(i + "\n")
warning_file.close()
```

```python

#info.txt file 
info_file= open("info.txt",'w')
for i in info_data:
    info_file.write(i + "\n")
info_file.close()
```

```python
# Load the log information into Pandas and perform frequency analysis.
import pandas as pd
```

```python
lines
```

```python
levels=[]
messages=[]
for i in lines:
    levels.append(i.strip().split("-")[0])
    messages.append(i.strip().split("-")[1])


    
```

```python
data= {"level": levels, "messages": messages}
df= pd.DataFrame(data)
```
