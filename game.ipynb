```python
##parent class Gamecharacter

class gamecharacter:
    def __init__(self, name, health, level):
        self.name= name
        self.health= health
        self.level= level

    def damage(self, damage_hp):
        self.health-= damage_hp
        if self.health<0:
            self.health=0
        print(f"{self.name} after taking damage {damage_hp},remaining health is {self.health}")  
```

```python
# child class warrior

class warrior(gamecharacter):
    def __init__(self, name, health, level, attack_type):
        super(). __init__(name, health,level)
        self.attack= attack_type

    def sword_attack(self, target):
        damage= 20* self.level
        print(f"{self.name} attacked {target.name} with {self.attack} for {damage} damage")
        target.damage(damage)
        
```

```python
class archer(gamecharacter):
    def __init__(self, name, health, level, attack_type):
        super(). __init__(name, health,level)
        self.attack= attack_type

    def archer_attack(self, target):
        damage= 15* self.level
        print(f"{self.name} attacked {target.name} with {self.attack} for {damage} damage")
        target.damage(damage) 
```

```python
class wizard(gamecharacter):
    def __init__(self, name, health, level, attack_type):
        super(). __init__(name, health,level)
        self.attack= attack_type

    def magic_attack(self, target):
        damage= 10* self.level
        print(f"{self.name} attacked {target.name} with {self.attack} for {damage} damage")
        target.damage(damage)   
```

```python
warrior1 = warrior("Thor", 100, 2, "Sword")
archer1 = archer("Robin", 80, 3, "Longbow")
wizard1 = wizard("Merlin", 70, 4, "Fireball")
```

```python
warrior1.sword_attack(archer1)
```
