```python
with open("sales.txt","w") as file:
    file.write(data)
```

```python
with open("sales.txt","r") as file:
    lines= file.readlines()
```

```python
lines
```

```python
import pandas as pd
```

```python
header= lines[0].strip().split(",")
```

```python
rows=[]
for i in lines[1:]:
    rows.append(i.strip().split(","))
```

```python
df= pd.DataFrame(data= rows, columns= header)
```

```python
df.head()
```

```python
df.info()
```

```python
df["price"]= df["price"].astype(float)
df["quantity"]= df["quantity"].astype(int)
```

```python
df.info()
```

```python
df["revenue"]=df["quantity"] * df["price"]
```

```python
# Find total revenue.
total_revenue= df["revenue"].sum()
print(total_revenue)
```

```python
 #Find average order value.
avg_order_value= df["revenue"].mean()
print(avg_order_value)


```

```python
# Find highest-value order.
highest_order_value= df["revenue"].max()
print(highest_order_value)

```

```python
# Find lowest-value order.
lowest_order_value= df["revenue"].min()
print(lowest_order_value)

```

```python
# Find revenue product-wise.
df.groupby("product")["revenue"].sum()

```

```python
# Find revenue city-wise.
df.groupby("city")["revenue"].sum()
```

```python
df.to_csv("sales_report.csv")
```
