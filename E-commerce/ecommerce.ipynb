```python
import pandas as pd
```

```python
df= pd.read_csv("product_dataset.csv")
```

```python
df.info()
```

```python
# Find total products.
print(df['product_id'].count())
```

```python
# Find unique categories.
df['category'].value_counts()
```

```python
# Find unique brands.
df['brand'].unique()
```

```python
df['brand'].value_counts()
```

```python
# Count products category-wise.
df["category"].value_counts()
```

```python
# Find average price.
print(df['price'].mean())

```

```python
# Find highest-priced product.
print(df['price'].max())
df[df['price']==df["price"].max()]
```

```python
# Find lowest-priced product.
df[df['price']==df["price"].min()]
```

```python
# Find products priced above ₹50,000.
df[df['price']>= 50000]

```

```python
# Find products with rating greater than 4.
df[df['rating']>4]
```

```python
# Sort products by price.
df.sort_values(by="price", ascending=False)
```

```python
# Sort products by rating.
df.sort_values(by="rating")
```

```python
# Find products with stock below 10.
df[df['stock']<10]
```

```python
# Create discount_amount.
df["discount_amount"]= df["discount"]*df["price"]*0.01
df.head()
```

```python
# Create final_price for 1 item
df["final_price"]=df['price']-df['discount_amount']
df.head()
```

```python
# Find average price by category.
df.groupby(by= "category")['price'].mean()
```

```python
# Find average rating by brand.
df.groupby(by= "brand")['rating'].mean()
```

```python
# Find maximum product price category-wise.
df.groupby(by="category")['price'].max()
```
