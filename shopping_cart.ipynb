```python
##product class containing Product ID Product Name Category Price and Available Stock

class product():
    def __init__(self, pID, pname,category, price, stock):
        self.product_ID= pID
        self.product_name= pname
        self.product_category= category
        self.product_price= price
        self.available_stock= stock


    def display_info(self):
        # Show product details in a neat format
        print(f"ID: {self.product_ID}")
        print(f"Name: {self.product_name}")
        print(f"Category: {self.product_category}")
        print(f"Price: ₹{self.product_price}")
        print(f"Available Stock: {self.available_stock}")
```

```python
class shopping_cart():

    def __init__(self):
        self.cart=[]

    def add_item(self, pname,quantity=1):
        if pname.available_stock>quantity:
            self.cart.append(pname)
            pname.available_stock= pname.available_stock-quantity
        print(f"{quantity}: {pname.product_name} added to cart. Stock available for {pname.product_name}  is {pname.available_stock}")

    def remove_item(self,pname,quantity=1):
        if pname in self.cart:
            self.cart.remove(pname)
            pname.available_stock=pname.available_stock+quantity
            print(f"{quantity}: {pname.product_name} removed to cart. Stock available for {pname.product_name}  is {pname.available_stock}")
        else:
            print(f"{pname.product_name} not added to cart")


    def display_cart(self):
        for i in self.cart:
            print(f"{i.product_name}: {i.product_price}")

    def total_cart_value(self):
        total= sum(i.product_price for i in self.cart)
        print(f"total cart value is {total}")
            
    
        

    
```
