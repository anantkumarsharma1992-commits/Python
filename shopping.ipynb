```python
products= {"Laptop":65000, "Phone":25000, "Mouse":800, "Keyboard":1200, "Monitor":9000, "Speaker":2500}

```

```python
order={"Laptop", "Phone", "Mouse", "Keyboard", "Chair", "Speaker"}
```

```python
for i in order:
    if i in products.keys():
        print("the product", i, "is available in the store")
    else:
        print("the product", i, "is not available in the store")
```

```python
laptop_quantity= int(input("Enter the quantity of Laptop you want to buy: "))
phone_quantity= int(input("Enter the quantity of Phone you want to buy: "))
mouse_quantity= int(input("Enter the quantity of Mouse you want to buy: "))
keyboard_quantity= int(input("Enter the quantity of Keyboard you want to buy: "))
monitor_quantity= int(input("Enter the quantity of Monitor you want to buy: "))
speaker_quantity= int(input("Enter the quantity of Speaker you want to buy: "))
```

```python
subtotal=laptop_quantity*products["Laptop"] + phone_quantity*products["Phone"] + mouse_quantity*products["Mouse"] + keyboard_quantity*products["Keyboard"] + monitor_quantity*products["Monitor"] + speaker_quantity*products["Speaker"]
print("The subtotal for the order is:", subtotal)
discount=0
if subtotal>10000:
    discount=subtotal*0.05
    print("The discount for the order is:", discount)
elif subtotal>30000:
    discount=subtotal*0.1
    print("The discount for the order is:", discount)
elif subtotal>60000:
    discount=subtotal*0.15
    print("The discount for the order is:", discount)


```

```python
print("***************Invoice****************")
print("product\t\tquantity\tprice")
print("-------\t\t--------\t-----")
print("Laptop\t\t", laptop_quantity, "\t\t",products["Laptop"])
print("Phone\t\t", phone_quantity, "\t\t",products["Phone"])
print("Mouse\t\t", mouse_quantity, "\t\t",products["Mouse"])
print("Keyboard\t", keyboard_quantity,"\t\t",products["Keyboard"])
print("Monitor\t\t", monitor_quantity, "\t\t",products["Monitor"])
print("Speaker\t\t", speaker_quantity, "\t\t",products["Speaker"])
print("------\t\t--------\t-----")
print("subtotal:\t\t\t",subtotal)
print("discount:\t\t\t",discount)
print("Final Amount:\t\t\t",subtotal-discount)    
```
